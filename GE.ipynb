```julia
function gauss_elim_solve(A)
    n = size(A, 1)
    @assert size(A,2) == n "A must be square"

    A = copy(A)

    for k in 1:n-1
        for i in k+1:n
            m = A[i, k] / A[k, k]
            A[i k:n] .-= m .* A[k k:n]
        end
        display(A)
    end
    return A
end
```

```julia
function get_clique_with_prob_p(n,p)
    L = zeros(Float64, n, n)
    for i in 1:n
        for j in i:n
            if rand() < p
                L[i,j] -= 1
                L[j,i] -= 1
                L[i,i] += 1
                L[j,j] += 1
            end
        end
    end
    L
end
```

```julia
function gremban_lift(L)
    n = size(L,1)
    M = zeros(Float64, 2*n, 2*n)
    for i in 1:n
        M[i,i] = L[i,i]
        M[i+n,i+n] = L[i,i]
        for j in i+1:n
            if L[i,j] > 0 # Move to A
                M[i,n+j] = -L[i,j]
                M[j,n+i] = -L[i,j]
                M[n+i,j] = -L[i,j]
                M[n+j,i] = -L[i,j]
            else
                M[i,j] = M[j,i] = L[i,j]
                M[i+n,j+n] = L[i,j]
                M[j+n,i+n] = L[i,j]
            end
        end
    end
    return M
end
```

```julia
function get_sddm_clique(n, indices_pairs, values)
    L = get_clique_with_prob_p(n,1)
    num_elems = size(indices_pairs,1)
    for pos_elem in 1:num_elems
        i = indices_pairs[pos_elem][1]
        j = indices_pairs[pos_elem][2]
        L[i, j] = values[pos_elem]
        L[j, i] = values[pos_elem]
        L[i, i] += abs(values[pos_elem]) - 1
        L[j, j] += abs(values[pos_elem]) - 1
    end
    return L
end
```

```julia
using Random
function create_laplacian(n, range_limit)
    L = zeros(n,n)
    for i in 1:n
        for j in i+1:n
        value= rand(0:range_limit)
        L[i, j] = -value
        L[j, i] = -value
        L[i, i] += abs(value)
        L[j, j] += abs(value)
    end
end
    return L
end
```

```julia
function check_thesis(L)
    n = size(L, 1)  # number of rows (assuming square matrix)
    Theory = true
    pseudo_inv = pinv(L)
    for i in 1:n
        for j in i+1:n
            for k in j+1:n
                if pseudo_inv[i,j] * pseudo_inv[i,k] < 0 && pseudo_inv[j,k] > 0
                    println("Theory not true")
                    print(pseudo_inv[i,j],pseudo_inv[i,k],pseudo_inv[j,k] )
                    Theory = false
                end
            end
        end
    end
    return Theory
end
```

```julia
function thesis_iteration(number_exp, size, range)
    for i in 1:number_exp
        L= create_laplacian(size,range)
        result= check_thesis(L)
        if result== false 
            return  L
        end 
    end 
    return true 
end
```

```julia
function create_disconnected_laplacian(n, range_limit)
    L = zeros(2*n,2*n)
    for i in 1:n
        for j in i+1:n
        value_1= rand(0:range_limit)
        value_2= rand(0:range_limit)
        L[i, j] = -value_1
        L[j, i] = -value_1
        L[i, i] += abs(value_1)
        L[j, j] += abs(value_1)
        L[i+n, j+n] = -value_2
        L[j+n, i+n] = -value_2
        L[i+n, i+n] += abs(value_2)
        L[j+n, j+n] += abs(value_2)
    end
end
    for i in 0:1 #change to a positve edge
        i_1=0
        j_1=0 
        while i_1== j_1 || L[i_1+i*n,j_1+i*n]==0
            i_1= rand(1:n)
            j_1= rand(1:n)
        end
        L[i_1+i*n,j_1+i*n]= -L[i_1+i*n,j_1+i*n]
        L[j_1+i*n,i_1+i*n]= L[i_1+i*n,j_1+i*n]
    end
    #Add connection edge
    i= rand(1:n)
    j= rand(1:n)
    value_1= rand(0:range_limit)
    L[i, j+n] = -value_1
    L[j+n, i] = -value_1
    L[i, i] += abs(value_1)
    L[j+n, j+n] += abs(value_1)

    return L
end
```

```julia
function effective_resistance(L_inv_new,a,b)
    e_a_b= zeros(size(L_inv_new,1))
    e_a_b[b,1]= 1 
    e_a_b[a,1]= -1
    r_eff= e_a_b'*L_inv_new*e_a_b
end
```

```julia
function check_triangle_inequality(L_new)
    print(size(L_new,1))
    n= size(L_new,1)
    L_inv= pinv(L_new)
    for i in 1:n 
        for j in i+1:n 
            for k in j+1:n 
                R_i_j=effective_resistance(L_inv,i,j)
                R_i_k= effective_resistance(L_inv,i,k)
                R_j_k= effective_resistance(L_inv,j,k)
                if R_i_j> R_i_k+  R_j_k
                    print(i,j,k)
                    print(R_i_j-R_i_k- R_j_k)
                    return false 
                elseif R_i_k> R_i_j+  R_j_k
                    print(i,j,k)
                    print(R_i_k-R_i_j-  R_j_k)
                    return false 
                elseif R_j_k> R_i_j+  R_i_k
                    print(i,j,k)
                    print(R_j_k-R_i_j-  R_i_k)
                    return false 
                end 
            end 
        end 
    end 
    return true 
end
```

```julia
function thesis_iteration_sing_conn(number_exp, size, range)
    for i in 1:number_exp
        L= create_disconnected_laplacian(size,range)
        result= check_triangle_inequality(L)
        if result== false 
            return  L
        end 
    end 
    return true 
end
```

```julia
function gaussian_and_check(L,index_elim)
    L_gauss=zeros(size(L,1)-1,size(L,1)-1)
    for k in 1:n-1
        for i in k+1:n
            w= L[index_elim,index_elim]
            m = -1*L[ index_elim, 1:end .!= index_elim]
            L_sub = L[1:end .!= index_elim, 1:end .!= index_elim]
            L_gauss= L_sub- 1/w* m*m'
            
        end
    end
    return L_gauss
 end
    
```
